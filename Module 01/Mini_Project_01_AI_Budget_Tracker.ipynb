## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
Write a Python program that:
1. Asks the user for their monthly income.
2. Asks the user to input up to three separate expenses, one at a time.
3. Calculates the total of the expenses and the remaining balance after deducting expenses from the income.
4. If the remaining balance is negative, it should print a warning message indicating that the budget is overspent and by how much.
5. Finally, it should display the monthly income, each expense, the total expenses, and the remaining balance.
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# [Paste your code here]

# 1. Ask the user for their monthly income.
while True:
    try:
        monthly_income = float(input("Enter your monthly income: "))
        if monthly_income < 0:
            print("Income cannot be negative. Please enter a positive number.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter a number for income.")

expenses = []
num_expenses = 0

# 2. Ask the user to input up to three separate expenses, one at a time.
while num_expenses < 3:
    expense_name = input(f"Enter name for expense {num_expenses + 1} (or press Enter to finish): ").strip()
    if not expense_name:
        break
    while True:
        try:
            expense_amount = float(input(f"Enter amount for {expense_name}: "))
            if expense_amount < 0:
                print("Expense cannot be negative. Please enter a positive number.")
            else:
                expenses.append({'name': expense_name, 'amount': expense_amount})
                num_expenses += 1
                break
        except ValueError:
            print("Invalid input. Please enter a number for the expense amount.")

# 3. Calculates the total of the expenses and the remaining balance.
total_expenses = sum(expense['amount'] for expense in expenses)
remaining_balance = monthly_income - total_expenses

# 4. If the remaining balance is negative, print a warning message.
if remaining_balance < 0:
    overspent_amount = abs(remaining_balance)
    print(f"\nWARNING: You have overspent your budget by ${overspent_amount:.2f}!")

# 5. Display the monthly income, each expense, the total expenses, and the remaining balance.
print("\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if expenses:
    for expense in expenses:
        print(f"  - {expense['name']}: ${expense['amount']:.2f}")
else:
    print("  (No expenses entered)")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Balance: ${remaining_balance:.2f}")
print("----------------------")
```

```python
# ENHANCED BUDGET TRACKER

# 1. Ask the user for their monthly income.
while True:
    try:
        monthly_income = float(input("Enter your monthly income: "))
        if monthly_income < 0:
            print("Income cannot be negative. Please enter a positive number.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter a number for income.")

expenses = []
num_expenses = 0

# 2. Ask the user to input up to three separate expenses, one at a time.
while num_expenses < 3:
    expense_name = input(f"Enter name for expense {num_expenses + 1} (or press Enter to finish): ").strip()
    if not expense_name:
        break
    while True:
        try:
            expense_amount = float(input(f"Enter amount for {expense_name}: "))
            if expense_amount < 0:
                print("Expense cannot be negative. Please enter a positive number.")
            else:
                expenses.append({'name': expense_name, 'amount': expense_amount})
                num_expenses += 1
                break
        except ValueError:
            print("Invalid input. Please enter a number for the expense amount.")

# 3. Calculates the total of the expenses and the remaining balance.
total_expenses = sum(expense['amount'] for expense in expenses)
remaining_balance = monthly_income - total_expenses

# 4. If the remaining balance is negative, print a warning message.
if remaining_balance < 0:
    overspent_amount = abs(remaining_balance)
    print(f"\nWARNING: You have overspent your budget by ${overspent_amount:.2f}!")

# 5. Display the monthly income, each expense, the total expenses, and the remaining balance.
print("\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if expenses:
    for expense in expenses:
        print(f"  - {expense['name']}: ${expense['amount']:.2f}")
else:
    print("  (No expenses entered)")
print(f"Total Expenses: ${total_expenses:.2f}")

# Enhancement: Calculate and display 'Percentage of Income Spent'
if monthly_income > 0:
    percentage_spent = (total_expenses / monthly_income) * 100
    print(f"Percentage of Income Spent: {percentage_spent:.2f}%")
else:
    print("Percentage of Income Spent: Cannot calculate percentage if income is zero.")

print(f"Remaining Balance: ${remaining_balance:.2f}")
print("----------------------")
```
