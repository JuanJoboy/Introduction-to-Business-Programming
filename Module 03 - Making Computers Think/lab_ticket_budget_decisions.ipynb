## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

Write a complete CLI Python program that guides a user through calculating and evaluating their monthly budget.

### Requirements & Specifications:

1. **User Details & Budget Initialization:**
   - Display welcome header: `"--- Welcome to your Smart Budget Checker! ---"`
   - Prompt for string inputs: `user_name` and `tracker_name`.
   - Prompt for `monthly_budget` and `current_money`. Use `while True` loops with `try...except ValueError` blocks to enforce positive numeric floats (`>= 0`).
   
2. **Budget vs. Available Funds Check:**
   - If `monthly_budget > current_money`, inform the user that their budget exceeds their available cash.
   - Ask if they want to adjust their budget via percentage (`yes`/`no`).
   - If `yes`: Prompt for a percentage between 0 and 100 (recommending 50%), calculate `monthly_budget = current_money * (percentage / 100)`, and display the updated budget.
   - If `no`: Reprompt for a manual `monthly_budget` that does not exceed `current_money`.
   - Display a setup confirmation greeting with `user_name`, `tracker_name`, `monthly_budget`, and `current_money`.

3. **Necessities Input Loop:**
   - Loop through necessities until the user enters `'done'` (case-insensitive).
   - Prompt for item name, cost per instance (positive float), and frequency per month (non-negative integer). Enforce validation loops with `try...except ValueError`.
   - Calculate item total (`cost * frequency`).
   - Granular Alert Checks: If the item total exceeds `monthly_budget` or `current_money`, display an alert showing the exact amount overspent.
   - Track total `necessities_cost`. If cumulative necessities exceed `monthly_budget`, print a cumulative alert warning with the exact excess dollar amount.

4. **Wants Input Loop:**
   - Loop through wants until the user enters `'done'` (case-insensitive).
   - Enforce input validation for item cost and frequency.
   - Granular Alert Checks: Alert if the item total exceeds `monthly_budget` or `current_money`.
   - Track total `wants_cost`.
   - Cumulative Alert Checks: Alert if total expenses (`necessities + wants`) exceed `monthly_budget`, or if `wants_cost` exceeds remaining funds after necessities.

5. **Budget Breakdown & Summary:**
   - Calculate:
     - `total_spent = necessities_cost + wants_cost`
     - `truly_spendable = monthly_budget - necessities_cost`
     - `misc_money = truly_spendable - wants_cost`
   - Print summary details: budget tracker name, total monthly budget, current money available, total expenses, and spendable amount after necessities (`max(0, truly_spendable)`).
   - Compute percentages for necessities, wants, and remaining misc relative to `monthly_budget` (handle division by zero if budget is 0).
   - Display percentage allocation messages. If `misc_money` is negative, print the exact over-budget amount.

6. **Final Tailored Advice:**
   - If `total_spent > monthly_budget`: Output exact overspend dollar amount and state: `"Consider cutting down on some expenses. I'm here to help."`
   - Else if `misc_money > 0`: Print an encouraging completion message with the user's name and tracker name.
   - Else: Print a encouragement message asking to find ways to balance the tracker.

```python
def budget_checker():
    print("\n--- Welcome to your Smart Budget Checker! ---")

    # Get user information
    user_name = input("What's your name? ")
    tracker_name = input("Give your budget tracker a name: ")

    while True:
        try:
            monthly_budget = float(input("How much money can you spend in a month in total? (e.g., 2000.00) "))
            if monthly_budget < 0:
                raise ValueError
            break
        except ValueError:
            print("Invalid input. Please enter a positive number for your monthly budget.")

    while True:
        try:
            current_money = float(input("How much money do you currently have? (e.g., 1500.00) "))
            if current_money < 0:
                raise ValueError
            break
        except ValueError:
            print("Invalid input. Please enter a positive number for your current money.")

    # New logic: Adjust monthly budget if it exceeds current money
    while monthly_budget > current_money:
        print(f"\nFriendly reminder: Your proposed monthly budget of ${monthly_budget:.2f} exceeds the money you currently have (${current_money:.2f}).")
        print("It's often a good idea to set a budget that's realistic for your current funds.")

        adjust_choice = input("Would you like to set your budget as a percentage of your current money? (yes/no): ").lower().strip()
        if adjust_choice == 'yes':
            while True:
                try:
                    percentage = float(input("Enter a percentage for your budget (e.g., 50 for 50%, recommended: 50%): "))
                    if not (0 <= percentage <= 100):
                        raise ValueError
                    monthly_budget = current_money * (percentage / 100)
                    print(f"Okay, your new monthly budget is set to {percentage:.0f}% of your current money: ${monthly_budget:.2f}.")
                    break # Exit percentage input loop
                except ValueError:
                    print("Invalid input. Please enter a percentage between 0 and 100.")
            if monthly_budget <= current_money: # Check if the adjusted budget is now valid
                break # Exit main budget adjustment loop
        else:
            while True:
                try:
                    new_monthly_budget = float(input(f"Please enter a new monthly budget (must be ${current_money:.2f} or less): "))
                    if not (0 <= new_monthly_budget <= current_money):
                        raise ValueError
                    monthly_budget = new_monthly_budget
                    break # Exit new budget input loop
                except ValueError:
                    print(f"Invalid input. Please enter a positive number no more than ${current_money:.2f}.")
            if monthly_budget <= current_money:
                break # Exit main budget adjustment loop

    print(f"\nHello {user_name}! Let's set up your '{tracker_name}'.")
    print(f"Your total monthly budget is ${monthly_budget:.2f}. You currently have ${current_money:.2f}.")

    necessities_cost = 0
    wants_cost = 0

    print("\n--- Let's list your NECESSITIES (e.g., rent, groceries, utilities) ---")
    while True:
        item_name = input("Enter a necessity (or 'done' to finish): ").strip()
        if item_name.lower() == 'done':
            break

        while True:
            try:
                item_cost = float(input(f"How much does {item_name} cost per instance? "))
                if item_cost < 0:
                    raise ValueError
                break
            except ValueError:
                print("Invalid input. Please enter a positive number for the cost.")

        while True:
            try:
                item_frequency = int(input(f"How many times will you pay for {item_name} this month? "))
                if item_frequency < 0: # Frequency should be non-negative
                    raise ValueError
                break
            except ValueError:
                print("Invalid input. Please enter a positive whole number for the frequency.")

        total_item_cost = item_cost * item_frequency

        # New granular overspending checks for individual items
        if total_item_cost > monthly_budget:
            print(f"  🚨 Alert: This '{item_name}' (${total_item_cost:.2f}) alone exceeds your total monthly budget (${monthly_budget:.2f}) by ${total_item_cost - monthly_budget:.2f}.")
        if total_item_cost > current_money:
            print(f"  🚨 Alert: This '{item_name}' (${total_item_cost:.2f}) alone exceeds your current available money (${current_money:.2f}) by ${total_item_cost - current_money:.2f}.")

        necessities_cost += total_item_cost

        # Cumulative necessities check
        if necessities_cost > monthly_budget:
            print(f"  🚨 Alert: Your total necessities (${necessities_cost:.2f}) have now exceeded your total monthly budget (${monthly_budget:.2f}) by ${necessities_cost - monthly_budget:.2f}.")

    print("\n--- Now, let's list your WANTS (e.g., entertainment, dining out, new gadgets) ---")
    while True:
        item_name = input("Enter a want (or 'done' to finish): ").strip()
        if item_name.lower() == 'done':
            break

        while True:
            try:
                item_cost = float(input(f"How much does {item_name} cost per instance? "))
                if item_cost < 0:
                    raise ValueError
                break
            except ValueError:
                print("Invalid input. Please enter a positive number for the cost.")

        while True:
            try:
                item_frequency = int(input(f"How many times will you pay for {item_name} this month? "))
                if item_frequency < 0:
                    raise ValueError
                break
            except ValueError:
                print("Invalid input. Please enter a positive whole number for the frequency.")

        total_item_cost = item_cost * item_frequency

        # New granular overspending checks for individual items
        if total_item_cost > monthly_budget:
            print(f"  🚨 Alert: This '{item_name}' (${total_item_cost:.2f}) alone exceeds your total monthly budget (${monthly_budget:.2f}) by ${total_item_cost - monthly_budget:.2f}.")
        if total_item_cost > current_money:
            print(f"  🚨 Alert: This '{item_name}' (${total_item_cost:.2f}) alone exceeds your current available money (${current_money:.2f}) by ${total_item_cost - current_money:.2f}.")

        wants_cost += total_item_cost

        # Cumulative wants check (after necessities)
        if (necessities_cost + wants_cost) > monthly_budget:
             print(f"  🚨 Alert: Your total expenses (necessities + wants) of ${necessities_cost + wants_cost:.2f} have now exceeded your monthly budget (${monthly_budget:.2f}) by ${necessities_cost + wants_cost - monthly_budget:.2f}.")
        elif wants_cost > (monthly_budget - necessities_cost) and (monthly_budget - necessities_cost) > 0:
            print(f"  🚨 Alert: Your 'wants' (${wants_cost:.2f}) have exceeded the money remaining after necessities (${monthly_budget - necessities_cost:.2f}) by ${wants_cost - (monthly_budget - necessities_cost):.2f}.")


    total_spent = necessities_cost + wants_cost
    truly_spendable = monthly_budget - necessities_cost
    misc_money = truly_spendable - wants_cost

    print("\n--- Your Budget Summary ---")
    print(f"Budget Tracker Name: {tracker_name}")
    print(f"Total Monthly Budget: ${monthly_budget:.2f}")
    print(f"Money Currently Available: ${current_money:.2f}")
    print(f"Total Expenses (Necessities + Wants): ${total_spent:.2f}")

    print(f"\nAmount you can truly spend after necessities: ${max(0, truly_spendable):.2f}")

    if monthly_budget > 0:
        percent_necessities = (necessities_cost / monthly_budget) * 100
        percent_wants = (wants_cost / monthly_budget) * 100
        percent_misc = (misc_money / monthly_budget) * 100
    else:
        percent_necessities = 0
        percent_wants = 0
        percent_misc = 0

    print(f"  - Necessities make up {percent_necessities:.2f}% of your total budget (${necessities_cost:.2f}).")
    print(f"  - Wants make up {percent_wants:.2f}% of your total budget (${wants_cost:.2f}).")

    if misc_money >= 0:
        print(f"  - You have {percent_misc:.2f}% (${misc_money:.2f}) remaining for miscellaneous spending! 🎉")
    else:
        print(f"  - You are over budget by ${abs(misc_money):.2f}! Time to re-evaluate. 😬")

    if total_spent > monthly_budget:
        print(f"\nOops! It looks like your planned expenses exceed your monthly budget by ${total_spent - monthly_budget:.2f}.")
        print("Consider cutting down on some expenses. I'm here to help.")
    elif misc_money > 0:
        print(f"\nGreat job, {user_name}! You're on track with your '{tracker_name}' budget. Keep up the smart spending! ✨")
    else:
         print(f"\nHey {user_name}, let's look for ways to balance your '{tracker_name}', so that you have some extra money. You got this! 💪")

# Run the budget checker
budget_checker()
```
