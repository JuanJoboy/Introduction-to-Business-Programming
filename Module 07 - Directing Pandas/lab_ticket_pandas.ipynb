## Task 1: total sales by region, highest first

Load the file into a DataFrame, clean whatever needs cleaning so the sale values are
real numbers, and produce the total `Total_Sale` for each `Region`, ordered from
highest to lowest.

Plan it in plain words in the cell below before you write the pandas. If you use an
assistant to help with a line, give it your plan and make sure you can explain what

My plan, in plain words:

- Clean the data by removing the `$` from the `Total_Sale` columns
- Add up all `Total_Sale` values for each region
- Order the table in descending order according to the `Total_Sale` column

```python
# Task 1: total sales by region, highest first.

import pandas as pd

def column_cleaner(df, column):
  df[column] = df[column].str.replace('$', '', regex=False) # regex=False makes the function treat the dollar sign as an actual dollar sign, rather than a regular expression phrase (normally it means 'end of line')
  df[column] = pd.to_numeric(df[column]) # Makes the column a number again instead of a string

def total_sum_of_column(df, aggregator_column, sum_column):
  sales = df.groupby(aggregator_column)[sum_column].sum() # Sums up the values in the sum_column according to what group it belongs to
  return sales.sort_values(ascending=False) # Sorts it in descending order

df = pd.read_csv('sales_data.csv') # Reads the csv and stores it in a DataFrame
column_cleaner(df, 'Total_Sale') # Cleans the Total_Sale column
total_sum_of_column(df, 'Region', 'Total_Sale') # Sums up the values according to what region they came from, sorts it, then displays the info
```
